```python
import pandas as pd

orders = pd.read_csv("../data/cleaned/orders_clean.csv")
customers = pd.read_csv("../data/cleaned/customers_clean.csv")
products = pd.read_csv("../data/cleaned/products_clean.csv")

print(orders.head())
```

```python
monthly_sales = orders.groupby("Month")["Sales"].sum()

monthly_sales
```

```python
plt.figure(figsize=(10,5))

plt.plot(monthly_sales)

plt.title("Monthly Sales")

plt.xlabel("Month")

plt.ylabel("Sales")

plt.show()
```

```python
monthly_profit = orders.groupby("Month")["Profit"].sum()

plt.figure(figsize=(10,5))

plt.bar(monthly_profit.index,monthly_profit.values)

plt.title("Monthly Profit")

plt.show()
```

```python
merged = orders.merge(products,on="ProductID")
```

```python
product_sales = merged.groupby("ProductName")["Sales"].sum()

product_sales
```

```python
product_sales.plot(kind="bar")

plt.show()
```

```python
customer_data = orders.merge(customers,on="CustomerID")
customer_sales = customer_data.groupby("CustomerName")["Sales"].sum()

customer_sales.sort_values(ascending=False)
```
